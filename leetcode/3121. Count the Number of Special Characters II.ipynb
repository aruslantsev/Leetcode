You are given a string word. A letter c is called special if it appears both in lowercase and uppercase in word, and every lowercase occurrence of c appears before the first uppercase occurrence of c.

Return the number of special letters in word.

 

Example 1:

Input: word = "aaAbcBC"

Output: 3

Explanation:

The special characters are 'a', 'b', and 'c'.

Example 2:

Input: word = "abc"

Output: 0

Explanation:

There are no special characters in word.

Example 3:

Input: word = "AbBCab"

Output: 0

Explanation:

There are no special characters in word.

 

Constraints:

    1 <= word.length <= 2 * 105
    word consists of only lowercase and uppercase English letters.


```python
class Solution:
    def numberOfSpecialChars(self, word: str) -> int:
        # state:
        # 0 - not seen
        # 1 - all previous only lowercase
        # 2 - lowercase than uppercase
        # 3 - invalid
        # transitioning paths
        # 0 -> 1 (first occurence is lowercase)
        # 1 -> 1 (next occurences are lowercase)
        # 0 -> 3 (first occurence is uppercase)
        # 1 -> 2 (current occurence is first uppercase)
        # 2 -> 2 (current occurence if not first uppercase)
        # 2 -> 3 (lowercase after uppercase)
        arr = [0 for _ in range(26)]
        for letter in word:
            if letter.islower():
                idx = ord(letter) - ord("a")
                if arr[idx] == 0:
                    arr[idx] = 1
                elif arr[idx] == 2:
                    arr[idx] = 3
            else:
                idx = ord(letter) - ord("A")
                if arr[idx] == 0:
                    arr[idx] = 3
                elif arr[idx] == 1:
                    arr[idx] = 2
        return len([e for e in arr if e == 2])
```
